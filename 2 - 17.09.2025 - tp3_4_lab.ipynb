```python
# Crear un rango de valores que exprese los niveles de importación como: Importante, medio, escazo" segun eliterio: sig. cr
#+++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++

# Si el vol. de importación es menor al 1° cuartil entonces la categoria es : "Escazo".
# Si el vol. de importación es mayor o igual al 1° cuartil  y menor que el 2° cuartil  la categoria es : "Medio".
# Si el vol. de importación es mayor al 2° cuartil inclusive  la categoria es : "importante".

# Crear una variable categórica 'Nivel_Impo' cuyas categorias son: "Escazo", "Medio", "importante".
#+++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++

#  Las condiciones son:
# q1= cuantil 25%
# q2= cuantil 50%

# Calculo de cuantiles
Q1=t3['Valor_CIF_USD'].quantile(0.25)
Q2=t3['Valor_CIF_USD'].quantile(0.50)

# Crea columna vacia
t3['Nivel_Impo']=''

#establece condicionales para col. vacia

t3.loc[t3['Valor_CIF_USD']<Q1,'Nivel_Impo'] ='Escaso'
t3.loc[(t3['Valor_CIF_USD']>=Q1)&(t3['Valor_CIF_USD']<Q2),'Nivel_Impo'] ='Medio'
t3.loc[t3['Valor_CIF_USD']>=Q2,'Nivel_Impo'] ='Importante'

# print(Q1, Q2)
t3
```

```python
# Normalizar datos: Z-score, método min-max (forma manual)
#++++++++++++++++++++++++++++++++++++++++++

# Z-score=xi-mu(x)/sd(x)

# Min-Max=xi-min(x)/Max(x)-Min(x)

# Calcula valores normalizados (z-score) de las variables: "Valor CIF (en Dólares)" y "Peso Neto (en Kilos)"
mu_cf=t3['Valor_CIF_USD'].mean()
sd_cf=t3['Valor_CIF_USD'].std()

t3['Z_VCIF']=(t3['Valor_CIF_USD']-mu_cf)/sd_cf

mu_pn=t3['Peso_Neto_KG'].mean()
sd_pn=t3['Peso_Neto_KG'].std()
t3['Z_PN']=(t3['Peso_Neto_KG']-mu_pn)/sd_pn

t3
```

```python
# Normalizar variables con  Standard Scaler / minmaxscaler # (forma automática)
#++++++++++++++++++++++++++++++++++++++++++++++++++++++++++

# StandardScaler is applied to each of the columns, where each value is subtracted with the mean of its column and divided by its standard deviation.
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler # xi-mu/sd

scaler=StandardScaler()
t3[['Z_auto_CIF','Z_auto_PN']]=scaler.fit_transform(t3[['Valor_CIF_USD','Peso_Neto_KG']])
t3
```

```python
scaler2=MinMaxScaler()
t3[['M_auto_CIF','M_auto_PN']]=scaler2.fit_transform(t3[['Valor_CIF_USD','Peso_Neto_KG']])
t3
```

```python
# Repetir una fila de un data frame por ej. 3 veces
rep=[3 if val==8580 else 1 for val in t3['Valor_CIF_USD']]
t3=t3.loc[np.repeat(t3.index.values,rep)]
print(len(t3))

t3
```

```python
# Modificar el valor de una celda de un df
#++++++++++++++++++++++++++++++++++++++++

# importante: las filas y columnas comienzan con la poscion 0,1,.......... El primer dato está en la posición "0"
# iloc "integer location": usa índices numéricos de posición para acceder a filas y columnas. 
#print(t2.iloc[1,0]) # muestra el dato de la 2da fila de la 1ra. columna.

t3.iloc[1,1]="rosaria" # toma la 2da.fila en la 2da.col.
t3.iloc[2,1]="ROsaRio" # toma la 3ra.fila en la 2da.col.
t3.iloc[3,1]="ROSARIO" # toma la 4ta.fila en la 2da.col.
t3.iloc[4,1]="RoSáRíO" # toma la 4ta.fila en la 2da.col.
t3

#"rosaria","ROsaRio","ROSARIO"
```

```python
#  Normalizar cadenas de caracteres mediante expresiones regulares REGEX 
#+++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++

# Reemplazar cadena de caracter  ### VER REGEX Regular Expression#
# Reemplaza variantes de "Rosario" (con o sin acentos) por "Rosario"
t3['aduana'] = t3['aduana'].str.replace(r'(?i)ros[a-záéíóúñ]{1,6}', 'Rosario', regex=True)

t3

# r': indica que es una cadena cruda  de caracteres (raw string).
# Los caracteres de escape: como \n ( salto de linea) o \t (espacio tabulador) no se interpretan.

#(?i):() Busca caract. ignorando los casps de mayúsculas/minúsculas).
# El patrón "ros" busca toodos los carac. que comiencen con "ros".
#[a-záéíóúñ]{1,6} busca de 1 a 6 caract. (incluye acentos y ñ).

# regex = True: Especifica que el primer argumento es una expresión regular, en este caso: r"(?i)ro[A-Z,a-z]{1,8}"
# case = False: La búsqueda no diferencia entre mayúsculas y minúsculas.
```

```python
# Modificar celdas del atributo "Pais"

t3.iloc[1,3]="BRASIL" # toma la 2da.fila en la 4ta.col.
t3.iloc[2,3]="BRAZIL" # toma la 3ra.fila en la 4ta.col.
t3.iloc[3,3]="BRAzil" # toma la 4ta.fila en la 4ta.col.
t3.iloc[4,3]="bRásil" # toma la 4ta.fila en la 4ta.col.
t3
```

```python
# Normalizar la cadena de carac. "Brasil" mediante expresiones regulares REGEX
t3['pais']=t3['pais'].str.replace(r'(?i)br[a-záéíóúñ]{1,6}', 'Brasil', regex=True)
t3
```

```python

# Crear datos vacios y NAN
#+++++++++++++++++++++++++++

# Reemplazar datos. El parametro de la f. replace es un diccionario {key:value,key:value}, inplace:True, reemplaza el dato en el Df original
t3['Valor_CIF_USD'].replace({8580:np.nan},inplace=True)
t3['Peso_Neto_KG'].replace({5.460000e+02:np.nan},inplace=True)
t3
```

```python
# Reemplazar datos nulos y vacios por la media o por  la moda
#+++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++

# Simple imputer se usa para completar los valores faltantes (NaN) en un conjunto de datos.
# Puede usar indicadores estadísticos (mean,median, mode) o un valor constante para reemplazar los valores faltantes.


from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values=np.nan, strategy='median') #np.nan
t3[['Valor_CIF_USD','Peso_Neto_KG']]=imp.fit_transform(t3[['Valor_CIF_USD','Peso_Neto_KG']])
t3
```
