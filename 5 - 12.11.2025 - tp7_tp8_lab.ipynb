```python
# G3: Crear una columna llamada Ratio_expo
#+++++++++++++++++++++++++++++++++++++++++
# El ratio muestra la proporción de exportaciones de cada país del Mercosur (Bra, Uru, Ven, Par, Uru) en relación al toal del bloque Mercosur.
# R=Pais/Mercosur

Expo1['Ratio_Expo_Bra']=Expo1['ica_exportaciones_brasil']/Expo1['ica_exportaciones_total_mercosur']
Expo1['Ratio_Expo_Par']=Expo1['ica_exportaciones_paraguay']/Expo1['ica_exportaciones_total_mercosur']
Expo1['Ratio_Expo_Uru']=Expo1['ica_exportaciones_uruguay']/Expo1['ica_exportaciones_total_mercosur']
Expo1['Ratio_Expo_Ven']=Expo1['ica_exportaciones_venezuela']/Expo1['ica_exportaciones_total_mercosur']
Expo1
```

```python
# G5-G6. Crear una variable categorica que clasifique el nivel total de exportaciones en el Mercosur como: bajo, medio ,alto, Excelente.
#++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++

# Es BAJO si el valor tot. de expo. es menor al primer cuartil.
# Es MEDIO si el valor tot. de expo. es mayor al primer cuartil y menor al segundo cuartil.
# Es ALTO si el valor tot. de expo. es superior al segundo cuartil y menor al tercer cuartil.
# Es EXCELENTE si el valor tot. de expo. es superior al segundo cuartil y menor al tercer cuartil.
```

```python
q1_em=Expo1['ica_exportaciones_total_mercosur'].quantile(0.25).round(4)
q2_em=Expo1['ica_exportaciones_total_mercosur'].quantile(0.50).round(4)
q3_em=Expo1['ica_exportaciones_total_mercosur'].quantile(0.75).round(4)

print("Q1:",q1_em,"Q2:",q2_em,"Q3:",q3_em)
```

```python
Expo1['Nivel_Expo_Msur']=''

Expo1.loc[Expo1['ica_exportaciones_total_mercosur']<q1_em,'Nivel_Expo_Msur']="Bajo"
Expo1.loc[(Expo1['ica_exportaciones_total_mercosur']>=q1_em)&(Expo1['ica_exportaciones_total_mercosur']<q2_em),'Nivel_Expo_Msur']="Medio"
Expo1.loc[(Expo1['ica_exportaciones_total_mercosur']>=q2_em)&(Expo1['ica_exportaciones_total_mercosur']<q3_em),'Nivel_Expo_Msur']="Alto"
Expo1.loc[Expo1['ica_exportaciones_total_mercosur']>=q3_em,'Nivel_Expo_Msur']="Excelente"

Expo1
```

```python
#Crear una funcion
#+++++++++++++++++

# Crear una variable llamada "Categoria" que asigne:
# 1 si el nivel de expo es "Bajo"
# 2 si el nivel de expo es "Medio"
# 3 si el nivel de expo es "Alto"
# 4 si el nivel de expo es "Superior"

# Definir el argumento de la funcion
Nivel_Expo=Expo1['Nivel_Expo_Msur']

# Crear una funcion cuyo argumento es el "Nivel de Expo. Mercosur"
def asignar_categoria(Nivel_Expo):  
    if Nivel_Expo=='Bajo':
        return "1"
    elif Nivel_Expo=='Medio':
        return "2"
    elif Nivel_Expo=='Alto':
        return "3"
    elif Nivel_Expo=='Excelente':
        return "4"
    else:
        return None
    
# Aplicar la funcion al data frame.
# Añade una nueva columna llamada"Categoria" al DataFrame Expo1 e invoca la función "asignar_categoria",
# donde asigna los valores 1,2,3,4  en base a cada nivel de expo. 
Expo1['Categoria'] = Expo1['Nivel_Expo_Msur'].apply(asignar_categoria)
Expo1
```
